```python
import pandas as pd
import json 
```

```python
with open("mmlu_scores_dove.json", "r") as f:
    json_file = json.load(f)
```

```python
MMLU_SIZE = 14042
ROBSUT_SCORE_THRESHOLD = 0.6
```

```python
json_file['101']
```

```python
full_result_file = {}
for i in range(MMLU_SIZE):
    if str(i) in json_file:
        if json_file[str(i)] > ROBSUT_SCORE_THRESHOLD:
            full_result_file[str(i)] = 1
        else:
            full_result_file[str(i)] = 0
    else:
        full_result_file[str(i)] = 1
```

```python
data = list(full_result_file.values())
with open("dove_full_result.json", "w") as f:
    json.dump(data, f, indent=2)
```
